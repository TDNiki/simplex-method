П2 

при делении столбец на столбец целевой столбец может иметь значения 0 или отрицательные. В итоге операции может возникнуть бесконечность или отрицательное число, что может навредить в поиске минимального значение из этого результата. Берем и присваиваем большое значение 

```python
def get_m1_matrix(A: np.array, f: np.array, b: np.array):
    """Получаем матрицу вида M"""
    part_1 = np.hstack([b, A])
    one_matrix = np.eye(A.shape[0])

    M = np.hstack((part_1, one_matrix))
    last_row = np.zeros((M.shape[1]))

    for index, val in enumerate(f):
        last_row[index + 1] = -val

    M = np.vstack((M, last_row))

    return M
    

def to_zero(m: np.array, new_matrix: np.array, target_column_index: int, target_row_index):


    for i in range(m.shape[0]):
        if i == target_row_index: continue

        new_matrix[i, :] = m[i, :] - m[i, :][target_column_index] * new_matrix[target_row_index, :]


def do(A: np.array, f: np.array, b: np.array, max_iter = 3, filter_indexs = set()):
    m = get_m1_matrix(A, f, b)
    iteration = 0

    

    while m[-1, :-6].min() < 0 and iteration <= max_iter: #

        min_f_column_index = m[-1, 1:-A.shape[0]].argmin() + 1 #индекс столбца, где наблюдается минимальное значение по -f; см П1

        zero_column_row = m[:-1, 0] / m[:-1, min_f_column_index]
        for i in range(zero_column_row.size):
            if zero_column_row[i] <= 0:  zero_column_row[i] =1000 # Смотреть П2
        min_zero_column_row_index = (zero_column_row).argmin() #индекс минимального значения в нулевом столбце
        

        target_row_transformed = m[min_zero_column_row_index, :] / m[:, min_f_column_index][min_zero_column_row_index]
        new_matrix = m.copy()
        new_matrix[min_zero_column_row_index, :] = target_row_transformed
        to_zero(m, new_matrix, min_f_column_index, min_zero_column_row_index)
        

        if not len(filter_indexs):
            print_matrix(m)
        elif iteration in filter_indexs:
            print_matrix(m)

        print('\n\n')

        m = new_matrix
        
        if m[:, 0].min() < 0:
            raise ValueError(f"iter: {iteration}; detect negative value {m[:, 0].min()} in {m[:, 0].argmin()} position; \n\n\
                             params:\n\
                             min_f_column_index = {min_f_column_index}\n\
                             zero_column_row = {zero_column_row}: \n\
                             min_zero_column_row_index = {min_zero_column_row_index}\n\
                             target_row_transformed = {target_row_transformed}\n\
                             ")
        iteration += 1


    return (iteration, m)
```
